# Verificação das ligações de nomes
Executar nesta pasta. Não escreve na BD. A fonte original e o ficheiro enriquecido são comparados sem alterar resultados ou identidades.

```python
import json
from pathlib import Path
base = json.loads(Path("dados_para_insercao_REVISAR.json").read_text())
linked = json.loads(Path("dados_para_insercao_COM_ALCUNHAS_REVISAR.json").read_text())
rows = json.loads(Path("correspondencias_nomes_alcunhas.json").read_text())["rows"]
assert len(rows) == base["summary"]["rosterEntries"]
assert len({(r["reportNumber"], r["teamId"], r["maId"]) for r in rows}) == len(rows)
for old, new in zip(base["matches"], linked["matches"]):
    assert old["reportNumber"] == new["reportNumber"]
    assert len(old["events"]) == len(new["events"])
    for old_event, new_event in zip(old["events"], new["events"]):
        assert all(new_event[k] == v for k, v in old_event.items())
    for side in ("home", "away"):
        assert old[side+"Score"] == new[side+"Score"]
        assert len(old["lineups"][side]) == len(new["lineups"][side])
        for old_player, new_player in zip(old["lineups"][side], new["lineups"][side]):
            assert all(new_player[k] == v for k, v in old_player.items())
print(linked["nameLinking"]["summary"])
```
